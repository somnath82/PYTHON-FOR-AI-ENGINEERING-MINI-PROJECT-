```python
# PYTHON FOR AI ENGINEERING MINI PROJECT
#  Project: Simple Document Search + Mock AI Assistant
# This combines:
# - functions
# - OOP
# - JSON
# - file handling
# - exceptions
# - modules
# - project-style structure
import json
from pathlib import Path
from dataclasses import dataclass


DATA_FILE = Path("documents.json")


@dataclass
class Document:
    id: int
    text: str
    category: str


class DocumentStore:
    def __init__(self):
        self.documents = []

    def add(self, document: Document):
        self.documents.append(document)

    def search(self, query: str):
        words = set(query.lower().split())
        results = []

        for doc in self.documents:
            text_words = set(doc.text.lower().split())
            score = len(words & text_words)

            if score > 0:
                results.append((score, doc))

        return sorted(results, key=lambda item: item[0], reverse=True)

    def save(self, path=DATA_FILE):
        data = [doc.__dict__ for doc in self.documents]
        path.write_text(json.dumps(data, indent=2), encoding="utf-8")

    def load(self, path=DATA_FILE):
        try:
            data = json.loads(path.read_text(encoding="utf-8"))
            self.documents = [Document(**item) for item in data]
        except FileNotFoundError:
            print("No database found. Starting empty.")
        except json.JSONDecodeError as exc:
            print("Invalid JSON:", exc)


def mock_ai_answer(question, context):
    return (
        "Mock AI answer:\n"
        f"Question: {question}\n"
        f"Relevant context: {context}"
    )


def main():
    store = DocumentStore()

    store.add(Document(
        1,
        "Python is a popular programming language for AI engineering.",
        "python"
    ))
    store.add(Document(
        2,
        "REST APIs allow applications to communicate over HTTP.",
        "backend"
    ))
    store.add(Document(
        3,
        "JSON is commonly used for structured API data.",
        "data"
    ))

    store.save()

    query = "Python AI"
    results = store.search(query)

    if not results:
        print("No relevant documents found.")
        return

    top_documents = [doc.text for score, doc in results[:2]]
    context = "\n".join(top_documents)

    print(mock_ai_answer(query, context))


if __name__ == "__main__":
    main()
```
